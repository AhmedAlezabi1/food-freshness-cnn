```python
import numpy as np

# # # Confusion matrix for the privileged group (female)

cm_priv = np.array([[45, 4],
                    [9, 40]])

# # # Confusion matrix for the unprivileged group (male)

cm_unpriv = np.array([[42, 7],
                      [14, 35]])
np.save('confusion_matrix_priv_female.npy', cm_priv)
np.save('confusion_matrix_unpriv_male.npy', cm_unpriv)
```

```python
import numpy as np

def load_confusion_matrices():
    cm_priv = np.load('./confusion_matrix_priv_female.npy')
    tn_priv = cm_priv[0][0]
    fp_priv = cm_priv[0][1]
    fn_priv = cm_priv[1][0]
    tp_priv = cm_priv[1][1]
    
    cm_unpriv = np.load('./confusion_matrix_unpriv_male.npy')
    tn_unpriv = cm_unpriv[0][0]
    fp_unpriv = cm_unpriv[0][1]
    fn_unpriv = cm_unpriv[1][0]
    tp_unpriv = cm_unpriv[1][1]
    
    return [[tn_priv, fp_priv, fn_priv, tp_priv],
            [tn_unpriv, fp_unpriv, fn_unpriv, tp_unpriv]]

def statistical_parity(conf_matrices):
    tn_priv, fp_priv, fn_priv, tp_priv = conf_matrices[0]
    tn_unpriv, fp_unpriv, fn_unpriv, tp_unpriv = conf_matrices[1]
    
    total_priv = tn_priv + fp_priv + fn_priv + tp_priv
    total_unpriv = tn_unpriv + fp_unpriv + fn_unpriv + tp_unpriv
    
    sp_priv = (tp_priv + fp_priv) / total_priv
    sp_unpriv = (tp_unpriv + fp_unpriv) / total_unpriv
    
    return [sp_priv, sp_unpriv, abs(sp_unpriv - sp_priv)]

def equal_opportunity(conf_matrices):
    tn_priv, fp_priv, fn_priv, tp_priv = conf_matrices[0]
    tn_unpriv, fp_unpriv, fn_unpriv, tp_unpriv = conf_matrices[1]
    
    tpr_priv = tp_priv / (tp_priv + fn_priv)
    tpr_unpriv = tp_unpriv / (tp_unpriv + fn_unpriv)
    
    return [tpr_priv, tpr_unpriv, abs(tpr_unpriv - tpr_priv)]

def false_positive_rate(conf_matrices):
    tn_priv, fp_priv, fn_priv, tp_priv = conf_matrices[0]
    tn_unpriv, fp_unpriv, fn_unpriv, tp_unpriv = conf_matrices[1]
    
    fpr_priv = fp_priv / (tn_priv + fp_priv)
    fpr_unpriv = fp_unpriv / (tn_unpriv + fp_unpriv)
    
    return [fpr_priv, fpr_unpriv, abs(fpr_unpriv - fpr_priv)]

confusion_matrices = load_confusion_matrices()

print("Statistical Parity:", statistical_parity(confusion_matrices))
print("Equal Opportunity:", equal_opportunity(confusion_matrices))
print("False Positive Rate:", false_positive_rate(confusion_matrices))
```
