```python

import numpy as np
import matplotlib.pyplot as plt
import skimage as ski
from scipy.signal import convolve2d

# Step 1: Creating the 3x3 Kernel for Vertical Edge Detection
array_1 = np.array([[1, 0, -1],
                    [1, 0, -1],
                    [1, 0, -1]])

# Step 2: Loading the BUas Image in Grayscale
buas_img = ski.io.imread("BUas_1.png", as_gray=True)

plt.imshow(buas_img, cmap="gray")
plt.title("Original BUas Image")
plt.show()
```

```python

# Step 3: Applying Convolution using the Vertical Edge Detection Kernel
conv_output_1 = convolve2d(buas_img, array_1)

plt.imshow(conv_output_1, cmap="gray")
plt.title("Vertical Edge Detection")
plt.show()
```

```python

# Step 4: Enhancing the Edges by Taking Absolute Values and Clipping
abs_conv_1 = np.absolute(conv_output_1)
clipped_result_1 = np.clip(abs_conv_1, 0, 1.0)

plt.imshow(clipped_result_1, cmap="gray")
plt.title("Enhanced Vertical Edge Detection")
plt.show()
```

```python

# Step 5: Creating the 3x3 Kernel for Horizontal Edge Detection
array_2 = np.array([[1, 1, 1],
                    [0, 0, 0],
                    [-1, -1, -1]])

# Step 6: Applying Convolution using the Horizontal Edge Detection Kernel
conv_output_2 = convolve2d(buas_img, array_2)

abs_conv_2 = np.absolute(conv_output_2)
clipped_result_2 = np.clip(abs_conv_2, 0, 1.0)

plt.imshow(clipped_result_2, cmap="gray")
plt.title("Horizontal Edge Detection")
plt.show()
```

### 2b Implement a sharpening filter. You can read about the Unsharp Masking technique here (see section Implementation for the kernel values). Apply the filter in BUas image.

```python
import numpy as np
import matplotlib.pyplot as plt
from skimage.io import imread
from skimage.color import rgb2gray
from scipy.signal import convolve2d

# Loadding the BUas image and convert to grayscale
image_path = "BUas_1.png"
buas_img = imread(image_path)
buas_gray = rgb2gray(buas_img)

# Defining the Unsharp Masking sharpening kernel
sharpen_kernel = np.array([[0, -1, 0],
                           [-1, 5, -1],
                           [0, -1, 0]])
```

```python

# Applying convolution with the sharpening kernel
sharpened_image = convolve2d(buas_gray, sharpen_kernel, mode='same', boundary='symm')

# Clip values to ensure they stay within valid grayscale range
sharpened_image = np.clip(sharpened_image, 0, 1.0)

# Displaying the original grayscale image
plt.figure(figsize=(10,5))
plt.subplot(1, 2, 1)
plt.imshow(buas_gray, cmap="gray")
plt.title("Original BUas Image (Grayscale)")
plt.axis("off")

# Displaying the sharpened image
plt.subplot(1, 2, 2)
plt.imshow(sharpened_image, cmap="gray")
plt.title("Sharpened BUas Image (Unsharp Masking)")
plt.axis("off")

plt.show()
```

 ### 2c Implement a Gaussian blur filter. You can read about Gaussian blur here (see section Sample Gaussian Matrix for the kernel values). Apply the filter in BUas image.

```python
# Import necessary libraries
import numpy as np
import matplotlib.pyplot as plt
from skimage.io import imread
from skimage.color import rgb2gray
from scipy.ndimage import convolve

# 1. Loading the BUas image
image_path = "BUas_1.png"
buas_img = imread(image_path)

# 2. Converting the image to grayscale
buas_gray = rgb2gray(buas_img)

# 3. Defining a 5x5 Gaussian kernel (Sample Gaussian Matrix)
gaussian_kernel = np.array([[1, 4, 7, 4, 1],
                            [4, 16, 26, 16, 4],
                            [7, 26, 41, 26, 7],
                            [4, 16, 26, 16, 4],
                            [1, 4, 7, 4, 1]])
```

```python

# 4. Normalizing the kernel so all values add up to 1
gaussian_kernel = gaussian_kernel / np.sum(gaussian_kernel)

# 5. Applying the Gaussian blur using convolution
blurred_image = convolve(buas_gray, gaussian_kernel, mode='constant', cval=0)

# 6. Displaying the blurred image
plt.imshow(blurred_image, cmap="gray")
plt.title("BUas Image with Gaussian Blur")
plt.axis("off")  
plt.show()
```
