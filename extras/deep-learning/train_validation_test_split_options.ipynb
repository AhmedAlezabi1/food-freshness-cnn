## 3a Using option 1 (i.e., using scikit-learn’s train_test_split once and use the validation_split parameter when you call the fit() method from Keras), perform the following tasks:

```python
from sklearn.datasets import make_classification

X, y = make_classification(n_samples=100, n_features=2, n_classes=2, random_state=42, n_clusters_per_class=1, n_redundant=0)
print(X.shape, y.shape)
```

## 1 Split the data using the function train_test_split;

```python
from sklearn.model_selection import train_test_split

# # # Splitting the dataset into 90% training and 10% test data
X_train, X_test, y_train, y_test = train_test_split(X, y, test_size=0.1, random_state=0)


print("Training set shape:", X_train.shape, y_train.shape)
print("Test set shape:", X_test.shape, y_test.shape)
```

## 2 Build and compile a simple MLP model;

```python
from tensorflow.keras.models import Sequential
from tensorflow.keras.layers import Dense

model = Sequential([
    Dense(10, activation="relu", input_shape=(2,)), 
    Dense(1, activation="sigmoid")  
])

# Compile the model
model.compile(optimizer="adam", loss="binary_crossentropy", metrics=["accuracy"])

```

## 3 Train the Model with a Validation Split

```python
history = model.fit(X_train, y_train,  epochs=20, batch_size=5, 
                    validation_split=0.1)
```

## 4 Evaluate the model.

```python
test_loss, test_accuracy = model.evaluate(X_test, y_test)

print(f"Test Loss: {test_loss}")
print(f"Test Accuracy: {test_accuracy}")
```

## 1 Split the data using the function train_test_split to create the train and test sets;

```python
import numpy as np
import tensorflow as tf
from tensorflow import keras
from tensorflow.keras.models import Sequential
from tensorflow.keras.layers import Dense
from sklearn.model_selection import train_test_split
from sklearn.datasets import make_classification
```

```python
# Step 1: First split - Train (90%) and Test (10%)
X_train, X_test, y_train, y_test = train_test_split(X, y, test_size=0.1, random_state=0)
```

## 2 Further split the training data using the function train_test_split to create the validation set;

```python

# Step 2: Second split - Train (80%) and Validation (10%) from the Training set
X_train, X_val, y_train, y_val = train_test_split(X_train, y_train, test_size=1/9, random_state=0)

# Print dataset sizes
print("Train set shape:", X_train.shape, y_train.shape)
print("Validation set shape:", X_val.shape, y_val.shape)
print("Test set shape:", X_test.shape, y_test.shape)
```

## 3 Train an MLP model with the validation data (use the validation_data parameter);

```python
model = Sequential([
    Dense(10, activation="relu", input_shape=(2,)),  
    Dense(1, activation="sigmoid")  
])


model.compile(optimizer="adam", loss="binary_crossentropy", metrics=["accuracy"])
```

```python

history = model.fit(X_train, y_train, epochs=20, batch_size=5, validation_data=(X_val, y_val))
```

## 4 Evaluate the model.

```python
# Evaluate the model on test set
test_loss, test_accuracy = model.evaluate(X_test, y_test)

print(f"Test Loss: {test_loss}")
print(f"Test Accuracy: {test_accuracy}")
```
