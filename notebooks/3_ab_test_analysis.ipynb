# A/B Test Analysis
We're going to conduct an Independent Samples T-test to analyse our A/B test. An Indepedent Samples T-test compares the differences between two means of two different samples.

Now, let's start analysing our gathered data! This block we won't dive into inferential statistics since it can get complex quite fast; we'll do that in Year 2, block A. For now, you just need to know that we need to test whether the data is normally distributed and whether the variances of both samples are equal. Otherwise, our statistical tests would not be valid. What we are going to statistically test is whether there is a statistically significant difference in the mean of a given variable for version A or B. 

### Test Assumptions

```python
from scipy import stats

# Questions we want to test
questions = [
    "I understood what I could use the app for",
    "I found the application intuitive to use.",
    "I thought the application was useful.",
    "I enjoyed using the application.",
    "Was the scanning process fast and easy to use?"
]

# Loop through each question to run tests
for q in questions:
    print(f"\n Checking assumptions for: {q}")

    # 1. Normality test (Shapiro-Wilk)
    normal_a = stats.shapiro(df_A[q])
    normal_b = stats.shapiro(df_B[q])

    # 2. Equal variances test (Levene's Test)
    homogeneity = stats.levene(df_A[q], df_B[q])

    # Print results
    print(f"Shapiro-Wilk p-value (Version A): {normal_a.pvalue:.4f}")
    print(f"Shapiro-Wilk p-value (Version B): {normal_b.pvalue:.4f}")
    print(" If either p-value is below 0.05, the data is NOT normally distributed.")

    print(f"Levene’s Test p-value: {homogeneity.pvalue:.4f}")
    print(" If p < 0.05, the variances are different (not equal).")
```

### Run the T-test

Here you will run the T-test. Based on the assumption check above, decide for each question whether you have to run the bootstrapped version or the regular version of the t-test. Run only the appropriate one, and copy the cells as many times as necessary. 

```python
from scipy import stats

questions = [
    "I understood what I could use the app for",
    "I found the application intuitive to use.",
    "I thought the application was useful.",
    "I enjoyed using the application.",
    "Was the scanning process fast and easy to use?"
]

for q in questions:
    print(f"\n T-test for: {q}")
    results = stats.ttest_ind(df_A[q], df_B[q], equal_var=True)
    print("t-statistic:", round(results.statistic, 4))
    print("p-value:", round(results.pvalue, 4))

    if results.pvalue < 0.05:
        print("Statistically significant difference")
    else:
        print(" Not statistically significant")
```

## Decision: Which Test to Use

### After running the assumption checks: All 5 questions failed the normality test. 2 questions also failed the equal variance test.

### Because the assumptions for a t-test were not met, I used the **bootstrapped test** for **all 5 questions**.This test does not assume normal distribution and is better suited for this data.

-----

```python
import numpy as np

# Simple bootstrapping function
def bootstrap_test(data1, data2, num_samples=10000):
    rng = np.random.default_rng()
    differences = []

    for _ in range(num_samples):
        sample1 = rng.choice(data1, size=len(data1), replace=True)
        sample2 = rng.choice(data2, size=len(data2), replace=True)
        diff = np.mean(sample1) - np.mean(sample2)
        differences.append(diff)

    observed_diff = np.mean(data1) - np.mean(data2)
    p_value = np.mean(np.abs(differences) >= np.abs(observed_diff))
    ci = np.percentile(differences, [2.5, 97.5])

    return observed_diff, ci, p_value

# Questions to test
questions = [
    "I understood what I could use the app for",
    "I found the application intuitive to use.",
    "I thought the application was useful.",
    "I enjoyed using the application.",
    "Was the scanning process fast and easy to use?"
]

# Run the bootstrapped test for each question
for q in questions:
    print(f"\nQuestion: {q}")
    diff, ci, p = bootstrap_test(df_A[q], df_B[q])
    
    print(f"Mean difference: {diff:.2f}")
    print(f"95% confidence interval: [{ci[0]:.2f}, {ci[1]:.2f}]")
    print(f"p-value: {p:.4f}")
    
    if p < 0.05:
        print(" Significant difference – the version change likely had an effect.")
    else:
        print(" Not significant – the change might not have had a real effect.")
```
